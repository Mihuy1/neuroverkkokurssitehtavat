## Pretrained Glove model

In this code cell the glove model txt file is opened and every word and vector of those words are stored in an dictionary `embedding_index`.

```python
import numpy as np

glove_path = 'models/glove.6B.100d.txt'
embedding_index = {}

with open(glove_path, encoding='utf8') as f:
    for line in f:
        values = line.strip().split()
        word = values[0]
        vector = np.asarray(values[1:], dtype='float32')
        embedding_index[word] = vector

print(f"Loaded {len(embedding_index)} word vectors")
```

Here the vectors man, woman and king are collected and the result_vector is created that is the calculated expression.

```python
vec_man = embedding_index["man"]
vec_woman = embedding_index["woman"]
vec_king = embedding_index["king"]

result_vector = vec_woman - vec_man + vec_king
```

Here there are 2 methods:
- `cosine_similarity(vec1, vec2)`:
    - which calculates the cosine similarity between 2 vectors.
- `find_most_similar(target_vec, top_n=10, ignore_words=set(["man", "woman", "king"]))`: 
    - which loops through each word and vector in the `embedding_index` dictionary and checks first if the word is in the `ignore_words` set, and then if it isn't, the cosine similarity is calculated, and the name and similarity cosine value is added to `similarities` dictionary. After that the dictionary is sorted and the top 10 are returned.

```python
from numpy.linalg import norm

def cosine_similarity(vec1, vec2):
    return np.dot(vec1, vec2) / (norm(vec1) * norm(vec2))


def find_most_similar(target_vec, top_n=10, ignore_words=set(["man", "woman", "king"])):
    similarities = {}
    for word, vec in embedding_index.items():
        if word in ignore_words:
            continue
        sim = cosine_similarity(target_vec, vec)
        similarities[word] = sim

    sorted_words = sorted(similarities.items(), key=lambda x: x[1], reverse=True)
    return sorted_words[:top_n]
```

Here the cosine similarity is calculated for the `result_vector` that was created earlier and then the word and score are printed afterwards.

```python
nearest = find_most_similar(result_vector)
print("Top similar words to vec('woman') - vec('man') + vec('king'):")
for word, score in nearest:
    print(f"{word}: {score:.4f}")
```

The output shows the top 10 words that are most similar to the vector created by the equation:
-  `vec('woman') - vec('man') + vec('king')`.

Seems like the result is a vector that is royal like a king but female instead of a male, for exaple, a queen (The closest).
